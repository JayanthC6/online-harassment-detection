# Sprint 1: Exploratory Data Analysis
Dataset: Davidson et al. Hate Speech and Offensive Language dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/davidson_hate_speech.csv')
df.head()
```

```python
# Class balance -- this is the single most important thing to look at first.
# 0 = hate_speech, 1 = offensive_language, 2 = neither
df['class'].value_counts().plot(kind='bar')
plt.title('Class distribution (note the imbalance)')
plt.show()
df['class'].value_counts(normalize=True)
```

```python
# Tweet length distribution
df['tweet_len'] = df['tweet'].str.len()
df['tweet_len'].hist(bins=40)
plt.title('Tweet character length')
plt.show()
```

```python
# Most common words per class after cleaning -- run preprocess.py's clean_text
# on each class subset and look at word frequencies. Useful for spotting which
# words the baseline TF-IDF model will lean on.
import sys
sys.path.append('../backend/ml')
from preprocess import preprocess
from collections import Counter

for cls, name in [(0, 'hate_speech'), (1, 'offensive_language'), (2, 'neither')]:
    words = ' '.join(df[df['class'] == cls]['tweet'].apply(preprocess)).split()
    print(name, Counter(words).most_common(10))
```
